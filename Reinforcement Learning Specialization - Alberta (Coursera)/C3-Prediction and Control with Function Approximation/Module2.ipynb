# Estimating value functions with supervised learning

## Moving to Parameterized Functions

Up until now, all has been tabular methods. That is, methods that store all values for each possible state. This tables are untrackable, so we need to approximate value functions in other ways.

### Linear Value Function Approximation

The value of each state is represented as a linear function of the weights multiplied by some fixed attributes of the states called **features** ($x_i$).

$$\hat{v}(s, \mathbf{w}) = \sum w_i x_i(s) = \langle \mathbf{w}, \mathbf{x}(s) \rangle$$

## Generalization and Discrimination

- **Generalization**:  Applying knowledge about specific situations to draw conclusions about a wider variety of situations
- **Discrimination**: The ability to make the value of two states different

### Categorizing Methods Based on Generalization and Discrimination

We want a learning methods that achieves good generalization and good discrimination.

### Summary
- Tabular representations have provided good discrimination, but no generalization
- Generalization is important for faster learning
- Having both generalization and discrimination is ideal

## The value error objective

In Supervised Learning, the important step is to find the objective to optimize.

### Idealized scenario

We have a sequence of states and actual values: $\{(S_1, V_{\pi}(S_1)), (S_2, V_{\pi}(S_2)), (S_3, V_{\pi}(S_3)),\cdots\}$. We want to use this data to find a parametrized function that closely aproximates $V_\pi$. We do this by adjusting the weights so that the output of the function is as close as possible to the actual values. 

<img src="./img/match.png" alt="drawing" width="500"/>

We need to specify some measure of how close the output is to the actual value.

### The Mean Squared Value Error Objective

- MSVE is a measure of how far the output is from the actual value, and is: $\bar{VE} = \sum_s \mu(s)[v_\pi (s) - \hat{v}(s, \mathbf{W})]^2$ 
- $\mu(s)$ is the probability of state $s$ in the current episode, that is, the importance of state $s$ in the current episode (how much we care about each state)
- A natural choose of $\mu$ is the fraction of time we spend in $s$ when following policy $\pi$ 
- $v_\pi (s)$ is the actual value of state $s$
- $\hat{v}(s, \mathbf{W})$ is the output of the function approximator

<img src="./img/mse.png" alt="drawing" width="500"/>

### Summary
- Policy evaluation under function approximation requires us to specify an objective 
- Mean Squiare Value Error is a common objective
