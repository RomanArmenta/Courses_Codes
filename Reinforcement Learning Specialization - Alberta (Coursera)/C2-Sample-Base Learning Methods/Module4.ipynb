# TD for Control

## GPI with with TD: Sarsa

We use GPI to find optimal policy, and TD to estimate value functions. Here, we use TD to do policy evaluation in GPI.

### Recall Generalized Policy Iteration

- Policy iteration
- GPI with MC (need the episode to end in order to update action value functions)

### From state-values to action-values

Our goal is to find an algorithm that can improve the policy after just one policy evaluation step: using TD.

### Sarsa algorithm

This algorithm uses a sequence $S_t, A_t, R_{t+1}, S_{t+1}, A_{t+1}$ where the actions are taken from the policy we want to improve $\pi$, and then the action-value function can be update following the rule (this algo is for policy evaluation): 

$$Q(S_t, A_t)\rightarrow Q(S_t, A_t) + \alpha (R_{t+1} + \gamma Q(S_{t+1}, A_{t+1}) - Q(S_t, A_t))$$

### Summary
- We can combine *generalized policy iteration* with *TD Learning* to find *improved policies*
- *Sarsa* is an *action-value* form of TD which combines these ideas

## How is Q-leaning off-policy?

We have only seen off-policy algos that use important sampling. Why Q-Learning is off-policy? Here we will:

- *Understand* how Q-Learning can be *off-policy* without using importance sampling
- Describe how learning on-policy or off-policy might affect performance in control

### Comparison with Sarsa

An agent estimates its value function according to expected returns under their target policy, but behaves according to their behavior policy. When the target policy and the behavior policy are the same, then it is on-policy leraning, otherwise, is off-policy algorithm.

- **Sarsa**: $Q(S_t, A_t) \leftarrow Q(S_t, A_t) + \alpha(R_{t+1} + \gamma Q(S_{t+1}, A_{t+1}) - Q(S_t, A_t))$ 
    - On-policy algorithm
    - The agent bootstraps off of the value of the action it's going to take next $A_{t+1}$, which is sampled from its behavior policy
- **Q-Learning**: $Q(S_t, A_t) \leftarrow Q(S_t, A_t) + \alpha(R_{t+1} + \gamma \max_{a'} Q(S_{t+1}, a') - Q(S_t, A_t))$
    - Off-policy algorithm
    - Bootstraps off of the largest action value in its next state $a'$ (is like sampling an action under an estimate of the optimal policy rather than the behavior policy)

### Behavior and target policies

- Target policy is the optimal policy, like, always
- Behavior policy is the one that we let to explore as much as possible all the state space of the environment

<img src="./img/off.png" alt="drawing" width="500"/>

- The difference between behavior policy and target policy shows that Q-Learning is off-policy

### No importance sampling

- If Q-Learning is off-policy, why there's not important sampling ratios? Because the agent is estimating action values with unknown policies.
- Does not need important sampling rations to correct for the difference in action selection.

<img src="./img/actval.png" alt="drawing" width="500"/>

- The action-value function $Q(S_{t+1}, a')$ represents the returns following each action in a given state
- The agents target policy $\pi(a'|S_{t+1})$ represents the probability of taking each action in a given state
- With these two, we can compute the expected return under its targer policy from any given state, in particular, the next state $\sum_{a'} \pi(a'|S_{t+1})Q(S_{t+1}, a') = \mathbb{E}_\pi [G_{t+1}|S_{t+1}]$
- Since the agent's target policy is greedy with respect to its action values, all non-maximum actions have prob. 0
- As a result, the expected return from that state is equal to a maximal action value from that state

### The cliff walking environment

<img src="./img/cliff.png" alt="drawing" width="500"/>

- **Q-Learning** 
    - learns the optimal value function, then it quickly learns that an optimal policy travels right alongside the cliff
    - Since his actions are taken from epsilon greedy behavior, traveling alongside the cliff occasionally results in falling off of the cliff
- **Sarsa**
    - Learns about his current policy, taking into account epsilon greedy selection
    - Accounting for occasional exploratory actions, it learns to take the longer but more reliable path
    - Is able to reach the goal more reliably

### Summary

- Q-learning is off-policy without using importance sampling
- Learning on-policy or off-policy may perform differently in control

## Expected Sarsa

<img src="./img/expsarsa.png" alt="drawing" width="500"/>

- Sarsa sample the next state given that take action $a$ at time $t$ from the environment dynamics $p(s', r| s, a)$, and sample the next action $A_{t+1}$ from its policy $\pi(a'|s')$
- Since the agent already knows the policy that it is following, instead of selecting $A_{t+1}$, we can just take the expected value $\sum_{a'}\pi(a'|S_{t+1})Q(S_{t+1}, a')$

### The expected Sarsa algorithm

$$Q(S_{t}, A_t) \leftarrow Q(S_t, A_t) + \alpha (R_{t+1} + \gamma \sum_{a'}\pi(a'|S_{t+1})Q(S_{t+1}, a') - Q(S_t, A_t))$$

- Almost the same as sarsa, but here the TD error uses the expected estimate of the next action value instead of a sample of the next action value
- Every time step, the agent has to average the next state-action values according to how likely they are under the policy
- Expected sarsa has a more stable update raget than sarsa
- Calculating the expected value over actions becomes expensive as the number of actions increases

### Summary

- The *expected sarsa* algo explicitly computes the expectation under its policy, which is more expensive than sampling but has lower variance

## Generality of Expected Sarsa

we've seen 3 TD algo for control:
- Sarsa
- Q-learning
- Expected sarsa

The sarsas both approximate the same Bellman equation.

### Off-Policy expected sarsa

- Expected sarsa: $Q(S_{t}, A_t) \leftarrow Q(S_t, A_t) + \alpha (R_{t+1} + \gamma \sum_{a'}\pi(a'|S_{t+1})Q(S_{t+1}, a') - Q(S_t, A_t))$
    - The next action is sampled from $A_{t+1}\sim\pi(a'|S_{t+1})$
    - The expectation over actions does not depend on the next action selected $A_{t+1}$
    - Actually, $\pi$ does not need to be equal to the behavior policy
    - Then, like Q-learning, can be used to learn off-policy without importance sampling
- Q-Learning is a special case of expected sarsa, since it is the same rule if we use greedy expected sarsa (greedify over $\pi$ instead of computing the expected value, i.e., the case of expected sarsa with $\pi = \pi_*$)

### Summary

- *Expected sarsa* uses the same thechnique as *Q-Learning* to learn *off-policy* without *importance sampling*
- Expected sarsa with a target policy that is greedy with respect to its action-value is exactly Q-learning
