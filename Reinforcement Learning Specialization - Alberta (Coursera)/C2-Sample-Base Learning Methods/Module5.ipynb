### Models store knowledge

- In this course, models store knowledge about the transition and reward dynamics

- Using simulated experience improves the sample efficiency, since using the models means fewer interactions of the agent with the world to came up with the same policy

### Types of models

- **Sample models**: Produces an actual outcome drawn from some underlying probabilities.
    - Computationally inexpensive
    - random outcomes can be produced according to a set of rules
    - For example, the problem of flipping 5 times a coin, this model only needs to flip 5 times a coin
- **Distribution models**: Completely specifies the likelihood or probability of every outcome
    - Contain more information, but can be difficult to specify and can become very large
    - For flipping 5 times a coin, it needs to capture the probabilities for all $2^5 = 32$ possible outcomes

<img src="./img/taipsof.png" alt="drawing" width="500"/>

### Summary

- Models are used to improve policies through a process called planning
- Sample models procedurally generate samples, without explicitly storing the probability of each outcome
- Distribution model contain a list of all outcomes and their probabilities

## Comparing Sample and Distribution Models

- Adventages
    - Sample models require less memory
    - Distribution models can be used to compute the exact expected outcome
    - Distribution models can be used to assess risk

## The Dyna Architecture

- In direct RL ,the agent learns from experience generated by interacting with the environment (Q-Learning)
- In planning, the agent uses simulated experience to learn and plan without interacting with the world
- Can we combine them? YES! DYNA B

### Combining Q-Learning and Q-Planning
- Q-learning uses interaction with the environment to get the experience
- Q-planning performs updates using simulated experience from the model
- The dyna architecture uses both

<img src="./img/dyna.png" alt="drawing" width="500"/>

- We have the usual environment and policy
- They generate a stream of experience (the "environment experience")
- Use this experience to perform direct RL updates
- The model learns from the environment experience
- The model is used to generate "model experience" (simulated experience)
- The "Search control" is used to control the simulated experience the model generates (What states the agent will plan from?)

### Summary

- The *dyna architecture* uses direct RL and planning updates to learn from both environment and model experience
- Direct RL updates use environment experience to improve a policy or value function
- Planning updates use model experience to improve a policy or value function

## What if the model is inaccurate?

Our goal is to:

- Identify ways in which models can be inaccurate
- Explain the effects of planning with an inaccurate model
- Describe how Dyna can plan successfully with an incomplete model

### How models can be inaccurate

- Models are inaccurate when transitions they store are different from transitions that happen in the environment
- When a model has missing transitions (because at the firsts episodes of learning the agent usually does not visit all of the action-state pairs) is called an *incomplete model*
- A model can be inaccurate if the environment changes

### Planning with an inaccurate model

- Depends on how the model is inaccurate
- If the agent experienced all possible action-state pairs in the environment, then the model knows all the possible transitions. Now, suppose that the environment changes (for example, an action-pair transition is no longer leading to the next state that the model stored), then the updates of the value function during the planning might change in the wrong direction since the model does not updated that transition, then is an inaccurate model (since planning improves the policy with respect to what the model thinks will happen, instead of what will actually happen in the environment)
- The model becomes outdated when the environment changes

### Summary

- Models can be inaccurate if:
    - they are incomplete
    - the environment changes
- Planning with an inaccurate model improves the policy or value function with respect to the model, and not the environment
- Dyna-Q can play with an incomplete model by only using sampling state-action pairs that have been previously visited

## In-depth with changing environments

We want to know if there is a way to still learn from planning in changin environments -> YOU CAN! Dyna-Q+ (exploration-explotation trade off thinguis)

### Exploration and exploitation for model accuracy

- When the model is inaccurate, planning will make the policy or value function worse with respect to the enviroment (the agent is interested in that the model is accurate)
- In general, an agent might want to double-check that all its model's transitions are correct, but double-check transitions with low valued actions will often lead to low reward
- In changing environment, the model could become inaccurate at any time

<img src="./img/exp.png" alt="drawing" width="500"/>

- The agent has to make a choose:
    - Explore to make sure it's model is accurate
    - Exploit the model to compute the optimal policy assuming that the model is correct
- In order to correct models of changing environment, the agent should explore places it has not been to in a while
- To encourage the agent revisit previous states periodically, we can add a bonus to the reward used in planning:

$$\text{New reward} = r + \underbrace{\color{green}{\kappa\sqrt{\tau}}}_{\text{Bonus}}$$

- Here, $r$ is the reward from the model, $\tau$ are the time steps since the transition was last visited in the environment (in the learning-Q thingui). $\tau$ is not updated in the planning loop (that would not be a real visit) and $\kappa$ is a small parameter that controls the influence of the bonus in the planning update

### The Dyna-Q+ algorithm

<img src="./img/dynaqmas.png" alt="drawing" width="500"/>

- By artificially increasing the reward this way in planning, we increase the value of state-action pairs it haven't benn visited recently
- If an action-state pair has not been visited in a long time, the $\tau$ grows and the bonus becomes bigger
- Eventually, planning will change the policy to go directly to that state due to the large bonus
- When visits the state, it might see a big reward (or dissapointed)
- Either way, the model will be updated to reflect the dynamics of the environment

### Dyna-Q vs Dyna-Q+ in a changing environment

- Gridworld with a wall
- All transitions reward +0 except goal +1
- Discount factor less-than-one to encourage fast goal reaching
- Epsilon greedy policy

<img src="./img/dainaquiu.png" alt="drawing" width="500"/>

- First, Dyna-Q+ lears faster the optimal path since encourages exploration more

<img src="./img/dainaquiuplas.png" alt="drawing" width="500"/>

- After giving a shortcut, the dyna-Q+ founds it faster faster than dyna-Q

### Summary

- Model accuracy produces a variant of the exploration-exploration-exploitation trade-off
- Dyna-Q+ uses exploration bonuses to explore the environment
