# Introduction to Monte Carlo methods

## What is Monte Carlo?

In RL, Monte Carlo allows to estimate values directly from experience, from sequences of states, actions and rewards. This is, to learn value function without the prior knowledge of the envirionment dynamics.

To use a pure Dynamic Programming approach, the agent needs to know the environment's *transition probabilities* (the $p$). Also, if they are a lot of possible transitions, the computation can be error-prone and tedious. 

When we don't know it, or we lack of computation, Monte Carlo methods can help. Monte Carlo methods estimate values by averaging over a large number of random samples.

In policy evaluation, MC methods first observe multiple returns from the same state, in order to approximate the expected return of that state. These returns can only be observed at the end of an episode, so we will focus on MC for episodic tasks.

## Using Monte Carlo for Prediction

Consider blackjack, where you need to get cards that sum close-to 21 but not exceeding. 

<img src="./img/bj.png" alt="drawing" width="500"/>

### Problem formulation

- *Undiscounted* MDP where each game of blackjack corresponds to an *episode*
- *Reward*: -1 for a loss, 0 for draw, and 1 for a win
- *Actions*: Hit or Stick
- *States* (200 in total):
    - Wheter the player has a usable ace (yes or not)
    - The sum of the player's cards (12-21)
    - The card the dealer shows (Ace-10)
- Cards are dealt from a deck with replacement (no tracking cards)
- *Policy*; Stop requesting cards when the plater's sum is 20 or 21.

### Implications of Monte Carlo learning

- We do not need to keep a *large model* of the environment
- We are estimating the value of an indivitdual state *independently* of the values of other states
- The computation needed to update value of each state does not depend on the size of the MDP (rather, in the length of the episode)

## Why does off-policy learning matter?

- *On-policy*: Improve and evaluate the policy being used to select actions. The agent learns about the policy used to generate the data. What does it mean to learn about a policy? In policy evaluation, we mean learning the value function. In control, we mean learning the optimal policy.
- *Off-policy*: Improve and evaluate a different policy from the one used to select actions. The policy that we are learning is off the policy that we are using for action selection.

Now, some other thinguis:

- Target policy $\pi(a|s)$: The one that the agent is learning. The value function that the agent is learning is based on the target policy. The value function that the agent is learning is based on the target policy (e.g., optimal policy).
- Behavior policy $b(a|s)$: The one that the agent takes actions from. Defines agent behavior (e.g., uniform random policy).

Why decoupling behavior from taget? Is another way to favorates explorations. There are a few other applications of off-policy learning (learning from demonstration, parallel learning), but facilitating exploration is the main motivation. The behavior policy must cover the target policy, that is, if $\pi(a|s)>0$ for some $a$ and $s$, then the behavior policy must be such that $b(a|s)>0$.

Off-policy learning is a generalization of on-policy learning, where $b = \pi$ in the last one.

### Summary

- *Off-policy learning* allows learning an *optimal policy* from *suboptimal behavior*
- The policy that we are *learning* is the *target policy*
- The policy that we are choosing *actions* from is the *behavior policy*

## Importance sampling

Here, we will see how to use importance sampling to estimate the expected value of a distribution using samples from a different distribution.

### Derivation of Important Sampling

Suppose we have a set of samples from a distribution $x\sim b$, but we want to compute $\mathbb{E}_\pi [X] = \sum_{x\in X}x\pi (x)$, that is, the expected value of the random variable $x$ respect to a different distribution $\pi$.

$\mathbb{E}_\pi [X] = \sum_{x\in X}x\pi (x)\\
\hspace{1.02cm}= \sum_{x\in X}x\pi(x)\frac{b(x)}{b(x)}\\
\hspace{1.02cm}= \sum_{x\in X}x\frac{pi(x)}{b(x)}b(x)\\
\hspace{1.02cm}= x\rho(x)b(x)\\
\hspace{1.02cm}= \mathbb{E}_b [X\rho(X)]\\
\hspace{1.02cm}\approx \frac{1}{n}\sum_{i=1}^n x_i \rho(x_i)$

where the samples $x_i \sim b$ are sampled by $b$. We defined $\rho(x) = \pi(x)/b(x)$ called the *importance sampling ratio*, and we have used that $\mathbb{E}[X] \approx (1/n)\sum_{i=1}^n x_i$, where $x_i$ are sampled from the distribution where the expected value is computed. Then, the importance sampling methods weights the importance of the samples sampled by $b$ respect to $\pi$.

### Summary

- Importance sampling uses samples from one probability distribution to estimate the expectation of a different distribution

## Off-Policy Monte Carlo prediction

Here, we will understand how to use importance sampling to correct returns and how to modify the MC prediction algorithm for off-policy learning.

### Recall: Monte Carlo

$v_\pi (s) = \mathbb{E}_\pi [G_t | S_t = s] \approx average(\text{Returns of samples of episodes following }\pi\text{ starting from }s)$

If we want to do this but following a different behavior $b$, we need to use importance sampling.

$v_\pi (s) = \mathbb{E}_\pi [G_t | S_t = s] \approx average(\rho * \text{Returns of samples of episodes following }b\text{ starting from }s)$

Here, $\rho$ is the probability of the trajectory under $\pi$ divided by the trajectory under $b$.

$$V_\pi (s) = \mathbb{E}_b [\rho G_t | S_t = s]$$
$$\rho = \frac{\mathbb{P}(\textbf{Trajectory under }\pi)}{\mathbb{P}(\textbf{Trajectory under }b)}$$

### Off-Policy Trajectories

Because of the Markov process property, we can write the probability of a trajectory $(A_t, S_{t+1}, A_{t+1},...,S_T)$ given that the initial state at time $t$ is $S_t$ and the agent takes actions from behavior $b$ as:

$$P(A_t, S_{t+1}, A_{t+1},...,S_T | S_t, A_{t_T})=\\
b(A_t|S_t)p(S_{t+1}|S_t, A_t)b(A_{t+1}|S_{t+1})p(S_{t+2}|S_{t+1}, A_{t+1})\cdots p(S_T | S_{T-1}, A_{T-1})\\
= \prod_{k=t}^{T-1}b(A_k|S_k)p(S_{k+1}|S_k, A_k)$$
$$\mathbb{P}(\textbf{Trajectory under }b) = \prod_{k=t}^{T-1}b(A_k|S_k)p(S_{k+1}|S_k, A_k)$$

$$\Rightarrow\rho_{t:T-1} = \prod_{k=t}^{T-1}\frac{\pi(A_k|S_k)}{b(A_k|S_k)}$$
