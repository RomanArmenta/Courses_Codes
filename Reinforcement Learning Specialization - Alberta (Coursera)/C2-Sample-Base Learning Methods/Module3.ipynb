# Introduction to Temporal Difference Learning

## What is Temporal Difference Learning?

From the textbook:

> If one had to identify one idea as central and novel to RL, it would be TD Learning. (Sutton)

### Review: Estimating Values from Returns

- In the prediction problem, our goal is to learn a value function that estimates the return startin from a given state: $v_\pi = \mathbb{E}_\pi [G_t | S_t = s]$. 

- Both TD and MC methods use experience to solve the prediction problem. Given some experience following a policy $\pi$, both methods update their estimate $V$ of $v_\pi$ for the nonterminal states $S_t$ occurring in that experience. 

- MC methods wait until the return following the bisit is known, then use that return as a target for $V(S_t)$. 

A simple every-visit Monte Carlo method suitable for nonstationary environment is:

$$V(S_t) \leftarrow V(S_t) + \alpha \left[G_t - V(S_t)\right]$$

- $G_t$ is the actual return following time $t$, and $\alpha$ is a constant step-size parameter. 

- Using a constant step-size means that this algorithm can form MC estimates without saving lists of returns (only looks at the immediate next step). 

- To compute $G_t$, we need to take samples of full trajectories. This means that we do not learn during the episode, but we want to be able to learn incrementally during the episode. So, we look for a new update target to achieve this goal.

### Bootstrapping

- The definition of discounted return is: 

$\hspace{7cm}G_t = R_{t+1} + \gamma R_{t+2} + \gamma^2 R_{t+3} + ...\\
\hspace{7.48cm}= R_{t+1} + \gamma G_{t+1}$

- The value of a state at time $t$ is the expected return at time $t$:

$\hspace{7cm}v_\pi (s) = \mathbb{E}_\pi [G_t | S_t = s]\\
\hspace{7.88cm} = \mathbb{E}_\pi [R_{t+1} + \gamma G_{t+1} | S_t = s]\\
\hspace{7.88cm}= \mathbb{E}_\pi [R_{t+1}|S_t = s] + \gamma\underbrace{\mathbb{E}_\pi [G_{t+1} | S_t = s]}_{v_\pi (S_{t+1})}\\
\hspace{7.88cm} = R_{t+1} + \gamma v_\pi (S_{t+1})$

### Temporal Difference

- Using the above recursion equation, we can write the return as $G_t \approx R_{t+1} + \gamma V(S_{t+1})$ and we get the following update rule:

$$V(S_t) \leftarrow V(S_t) + \alpha \underbrace{[\overbrace{R_{t+1} + \gamma V(S_{t+1})}^{\text{TD target}} - V(S_t)]}_{\text{TD error }\delta_t}$$

- We can think of the value of the next state as a stand-in for the return until the end of the episode, so we don't have to wait until the end of the episode.

- In DP, we update $V(S)$ toward the value of all possible next state, using an expectation over all possible next states, needing a model $p$ of the environment to compute this expectation.

- In TD, we only neeed the next state. We can get directly from the environment without a model.

### Summary

- *TD Learning* - a way to incrementally estimate the return through *bootstrapping*

- *TD-error*: $\delta_t = R_{t+1} + \gamma V(S_{t+1}) - V(S_{t})$

- The *TD(0)* algorithm

## Advantages of TD Learning

- TD combines key ideas from DP and MC. Like DP, TD bootstrap. Like MC, TD learns from experience. The combination is more powerful that the sum of its parts.

- In TD, we can learn online without waiting for the episode to end like in MC

- TD methods do not require a model of the environment, they learn from experience

- Unlike MC, TD can update the values on every step. Bootstrapping allows us to update the estimates based on other estimates

- TD asymtotically converges to the correct predictions faster than MC methods (usually).
