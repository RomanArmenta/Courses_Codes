## 3.5 - Policies and Value Functions

Most of RL algo involve estimating *value functions* (functions of states or of state-action pairs that estimate *how good* it is for the agent to be in a given state (or how good is to perform a given action in a given state)). Value functions are defined with respect to particular ways of acting, called *policies*.

*Policy* is a mapping from states to probabilities of selecting each possible action. If an agent is following policy $\pi$ at time $t$, then $\pi(a|s)$ is the prob. that $A_t = a$ if $S_t = s$.

The *value function* of a state $s$ under a polivy $\pi$ denoted $v_\pi (s)$ is the expected return when starting in $s$ and following $\pi$ thereafter. For MDPs, 

$$v_\pi (s) = \mathbb{E}_\pi [G_t | S_t = s] = \mathbb{E}_\pi \left[\sum_{k=0}^\infty \gamma^k R_{t+k+1} | S_t = s\right], \text{ for all }s\in\mathcal{S}.$$

This function is called *state-value function for policy* $\pi$. We can define the value of taking action $a$ in state $s$ under a policy $\pi$ asi,

$$q_\pi (s, a) = \mathbb{E}_\pi [G_t | S_t = s, A_t = a] = \mathbb{E}_\pi \left[\sum_{k=0}^\infty \gamma^k R_{t+k+1} | S_t = s, A_t = a\right].$$

This function is called *action-value function for policy* $\pi$.

For any policy $\pi$ and any state $s$, the following consistency condition holds between the value of $s$ and the value of its possible successor states:

$\hspace{9cm}v_\pi (s) = \mathbb{E}_\pi [G_t | S_t = s]\\
\hspace{9.88cm} = \mathbb{E}_\pi [R_{t+1} + \gamma G_{t+1}| S_t = s]\\
\hspace{9.88cm} = \sum_a \pi(a|s)\sum_{s'}\sum_r p(s', r | s, a) \left[r + \gamma \mathbb{E}_\pi [G_{t+1}|S_{t+1}=s']\right]\\
\hspace{9.88cm} = \sum_a \pi(a|s)\sum_{s', r}p(s', r | s, a)\left[r + \gamma v_\pi (s')\right], \text{ for all }s\in\mathcal{S}$

where $a\in\mathcal{A}(s)$, $s' \in \mathcal{S}$ and $r\in\mathcal{R}$. This equation is called *Bellman equiation* for $v_\pi$. It expresses a relationship between the value of a state and the values of its successor states.

In the image, each open circle represents a state and each solid circle represents a state-action pair. Starting from state $s$, the agent could take any of some set of actions based on policy $\pi$. The Bellman equiation averages over all possibilities, weighting each of its probability of occurring. It states that the value of the start state must equal all the (discounted) value of the expected next state, plus the reward expected along the way.

## 3.6 - Optimal Policies and Optimal Value Functions

RL means to find a policy that achieves a lot of reward over the long run. For finite MDPs, an *optimal policy* (or policies) is the one that $v_\pi (s)$ is maximum for all $s\in\mathcal{S}$, denoted by $\pi_*$.  The *optimal state-value function* is given by,

$$v_* (s) = \max_\pi v_\pi (s), \text{ for all }s\in\mathcal{S}.$$

Optimal policies also share the same *optimal action-value function*:

$$q_* (s, a) = \max_\pi q_\pi(s, a), \text{ for all }s\in\mathcal{S} \text{, }a\in\mathcal{A}$$

$$q_* (s, a) = \mathbb{E}[R_{t+1} + \gamma v_* (S_{t+1})|S_t = s, A_t = a].$$

The *Bellman optimaly equation* expresses the fact that the value of a state under an optimal policy must be equal the expected return for the best action from that state:

$\hspace{9cm}v_* (s) = \max_{a\in\mathcal{A}(s)} q_{\pi_*}(s, a) \\
\hspace{9.85cm} = \max_a \mathbb{E}_{\pi_*}[G_t | S_t = s, A_t = a]\\
\hspace{9.85cm} = \max_a \mathbb{E}_{\pi_*}[R_{t+1} + \gamma G_{t+1} | S_t = s, A_t = a]\\
\hspace{9.85cm} = \max_a \mathbf{E}[R_{t+1} + \gamma v_* (S_{t+1})|S_t = s, A_t = a]\\
\hspace{9.85cm} = \max_a \sum_{s', r}p(s', r | s, a)[r + \gamma v_* (s')]$

The Bellman optimality equation for $q_*$ is:

$\hspace{9cm}q_* (s, a) = \mathbb{E}\left[R_{t+1} + \gamma \max_{a'}q_* (S_{t+1}, a')\mid S_t = s, A_t = a\right]\\
\hspace{10.18cm} = \sum_{s', r}p(s', r|s, a)\left[r + \gamma \max_{a'}q_* (s', a')\right]$
