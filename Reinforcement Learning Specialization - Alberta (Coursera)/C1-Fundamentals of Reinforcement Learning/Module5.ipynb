Uses Bellman eqs with the knowledge of $p$ (environment). Classical dynamical programming no use $p$, but a model of the environment.

## Summary

- *Policy evaluation* is the task of determining the state-value function $v_\pi$ for a particular $\pi$
- *Control* is the task of improving an existing policy
- *Dynamic programming* techniques can be used to solve both these tasks, if we have access to the *dynamics function* $p$

# Iterative Policy Evaluation (IPE)

dynamic algo are obtained by transforming the Bellman equations into update rules. The first of this algos is IPE.

Bellman eq: $v_\pi (s) = \sum_a \pi(a|s)\sum_{s'}\sum_r p(s', r|s, a)[r + \gamma v_\pi (s')]$

Use it as update rule:

$v_{k+1} (s) = \sum_a \pi(a|s)\sum_{s'}\sum_r p(s', r|s, a)[r + \gamma v_{k} (s')]$

This can be done because $v_\pi$ is the unique solution of the Bellman equation.

<img src="./img/IPE.png" alt="drawing" width="500"/>

## Summary

- We can turn the Bellman equation into an *update rule* to *iteratively* compute value functions

# Policy improvement

Given $v_*$ we find the optimal policy by greedy action.

<img src="./img/GREEDY.png" alt="drawing" width="500"/>

If we greedy action some value function $v_\pi$ as in the following image, the new policy must be different than $\pi$. If this "greedyfication" doeesnt change $\pi$, then the policy is already greedy with respect to its own value function, i.e., $v_\pi$ obeys Bellman optimality equation i.e. $\pi$ is optimal.

<img src="./img/coso.png" alt="drawing" width="500"/>

The new policy is a strict improvement over $\pi$ unless $\pi$ is already optimal -> This is a consequence of the *policy improvement theorem*.

## Policy Improvement Theorem

If $q_\pi (s, \pi'(s))\ge q_\pi (s, \pi(s))$ for all $s\in\mathcal{S}$, then $\pi' \ge \pi$. 

If $q_\pi (s, \pi'(s))> q_\pi (s, \pi(s))$ for at least one $s\in\mathcal{S}$, then $\pi' > \pi$. 

## Summary

- The *Policy Improvement Theorem* tells us that a greedfield policy is a strict improvement
- Use the value function under a given policy, to produce a strictly better policy

# Policy iteration

We can find the optimal policy by iterating the value function, called *policy iteration algorithm*. The policy improvement theorem tells us that we can construct a strictly better policy by actin greedly to respect to the value function of a given policy, unless the given policy is already optimal.

## Policy iteration

Consists in two steps repiting over and over: evaluation and improvement. First, evaluate our current policy $\pi_1 \rightarrow v_{\pi_1}$ (the value function accurately reflects the value of $\pi_1$).  Then, use $v_{\pi_1}$ to produce a strictly better greedy policy $\pi_2$. This new policy is greedy to respect the value function $v_{\pi_1}$, but this value function does not longer accurately reflects the value of $v_{\pi_2}$. The go on.

1. Evaluation: Given $\pi_1$, we can use iterative policy evaluation to construct $v_{pi_1}$.
2. Improvement: Greedify to respect $v_{\pi_1}$ to obtain a better policy, $\pi_2$, $\pi_2 > \pi_1$.

# Flexibility of the Policy Iteration Framework

## Generalized Policy iteration

Refer all the ways we can iteratively use policy evaluation and policy improvement.

1. Value iteration: we still sweep over all the states and greedify with respect to the current value function, but not run policy evaluation to completion, but only one sweep over all states, and greedify again. **(Sweep: just one correction of the value function over all states, not iteratively)**

<img src="./img/valueit.png" alt="drawing" width="500"/>

The difference is that we not follow any policy $\pi$ in the update of the value function, but the greedy policy. This still converges to $v_*$ in the limit. At the end, we can recover the optimal policy be greedly (argmax) selectin actions given the converged optimal value function. 

The methods that sweeps all the state space on each iteration are called *synchronous DP* (Value iteration, policy iteration). 

Asynchronous DP algo update the value of states in any order, not performing systematic sweeps.

## Summary

- *Value iteration* allow us to combine policy evaluation and improvement into a single update
- *Asynchronous* DP methods give us the freedom to update states in any order
- *Generalized Policy Iteration* unifies classical DP methods, value iteration and asynchronous DP.

# Efficiency of DP

- Policy iteration: Polinomial time in $|\mathcal{S}|$ and $|\mathcal{A}|$, exponentially faster that brute-force search ($|\mathcal{A}|^{|\mathcal{S}|}$ policies).

- The curse of dimensionality: the size of the state space grows exponentially as the number of relevant features increases (not an issue with DP, but an inherent complexity of the problem)

## Summary

- Bootstrapping (use known value functions to improve itself) can save us from perorming a huge amount of unnecessary work
