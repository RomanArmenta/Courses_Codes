## Chapter 3: Finite MArkov Decision Processes
The formal problem of finite Markov decision processes, or finite MDP, involves evaluative feedback, as in bandits, but also an associative aspect—choosing different actions in different situations.


### 3.1 - Agent-Environment Interface

The learner and decision maker is called the *agent*. The thing it interacts with, comprising everything outside the agent, is called the *environment*. More specifically, the agent and environment interact at each of a sequence of discrete time steps, $t = 0,1,2,....$. At each time step, the agent receives some representation of the environment's *state*, $S_t \in \mathcal{S}$, and on that basis selects an *action*, $A_t \in \mathcal{A}(s)$. One time step later, in part as a consequence of its action, the agent recieves a numerical *reward*, $R_{t+1}\in \mathcal{R} \in \mathbb{R}$, and finds itself in a new state, $S_{t+1}$. In the final run, the goal of the agent is to maximize the total reward for all time steps where it interacts with the environment.

In a *finite* MDP, the sets of states, actions and rewards ($\mathcal{S}$, $\mathcal{A}$ and $\mathcal{R}$) all have a finite number of elements. In this case, $R_t$ and $S_t$ have well defined discrete probability distributions dependent only on the preceding state and action. That is, for particular values $s'\in\mathcal{S}$ and $r\in\mathcal{R}$, there is a probability of those values occurring at time $t$, given particular values of the preceding state and action:

$$p(s', r|s, a) = \text{Pr} \{S_t=s', R_t = r | S_{t-1} = s, A_{t-1} = a\}.$$

The function $p$ defines the *dynamics* of the MDP.

In a *Markov* decision process, the probabilitiesw given by *p* completely characterize the environment's dynamics, since the probability of each possible value $S_t$ and $R_t$ depends only on the immediately preceding state and action $S_{t-1}$ and $A_{t-1}$. This is best viewed as a restriction not on the decision process, but on the *state*, since must include information about all aspects of the past agent-environment interaction that make a difference for the future.

From that distribution, we can get the *state-transition probabilities*:

$$p(s'|s, a) = \text{Pr}(S_t = s' | S_{t-1} = s, A_{t-1} = a) = \sum_{r\in \mathcal{R}}p(s', r|s, a).$$

The *expected rewards for state-action*:

$$r(s, a) = \mathbb{E}[R_t | S_{t-1} = s, A_{t-1} = a] = \sum_{r\in\mathcal{R}}r \sum_{s'\in\mathcal{S}}p(s', r| s, a).$$

The *expected rewards for state-action-next state* triples:

$$r(s, a, s') = \mathbb{E}[R_t | S_{t-1} = s, A_{t-1} = a, S_t = s'] = \sum_{r\in \mathcal{R}}r\frac{p(s', r | s, a)}{p(s' | s, a)}.$$

### 3.2 - Goals and Rewards

In RL, the purpose or goal of the agend is formalized in terms of a special signal, called *reward*, passing from the environment to the agent.

### 3.3 - Returns and Episodes

In order to learn, the agent's goal is to maximize the cumulative reward it receives in the long run. In general, we seek to maximize the *expected return* G_t, some specific function of the reward sequence. In the simplest case,

$$G_t = R_{t+1} + R_{t+2} + \cdots + R_{T},$$

where $T$ is a final time step. This approach makes sense when the agent-environment interaction breaks naturally into subsequences, which we call *episodes* (play a game, trips through a maze, anyrepeated interaction). Each episode ends in a *terminal state*, followed by a reset to a standard starting state or to a sample from a standard distribution of starting states. This type of tasks are called *episodic tasks*. Here we sometimes need to distinguish the set of all nonterminal states $\mathcal{S}$ from the set of all states plus the terminal states $\mathcal{S}^+$.

In many cases, the agent-environment interaction does not break naturally into identifiable episodes, but goes on continually without limit (on-going process-control task, application to a robot with a long life span, etc), called *continuing tasks*. Then, use the concept of *discounting*, where the agent tries to select actions so that the sum of the discounted rewards it recieves over the future is maximized. In particular, it chooses $A_t$ to maximize the expected *discounted return*,

$G_t = R_{t+1} + \gamma R_{t+2} + \gamma^2 R_{t+3} +\cdots = \sum_{k=0}^\infty \gamma^k R_{t+k+1},$

where $0\le\gamma\le 1$ is the *discount rate*, and determines the present value of future rewards.

Returns at successive time steps are related to each other,

$$G_t = R_{t+1} + \gamma G_{t+1}.$$

This works for all steps $t<T$. If $\gamma<1$, then $G_t$ is finite if the reward is nonzero and constant.
